```python
# NeuroScan AI — Brain Tumor MRI Dataset Loading + Preprocessing
# Google Colab: Run this entire script in ONE cell.
# Supports either an extracted Kaggle dataset folder or a downloaded ZIP file.

import os
import zipfile
import random
import hashlib
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image, ImageFile
from sklearn.model_selection import train_test_split

import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

# ---------------------------- CONFIGURATION ----------------------------
SEED = 42
IMAGE_SIZE = 224
BATCH_SIZE = 32
NUM_WORKERS = 2
VAL_SIZE = 0.15

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

ImageFile.LOAD_TRUNCATED_IMAGES = False

# ImageNet normalization: suitable for transfer-learning models such as EfficientNet/ResNet.
IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD  = (0.229, 0.224, 0.225)

# ---------------------------- FIND DATASET -----------------------------
# If automatic discovery fails, set DATASET_PATH to your extracted folder or ZIP path.
DATASET_PATH = ""  # Example: "/content/brain-tumor-mri-dataset"
SEARCH_ROOTS = [Path("/content"), Path("/kaggle/input")]

CLASS_ALIASES = {
    "glioma": "glioma",
    "glioma_tumor": "glioma",
    "glioma tumor": "glioma",
    "meningioma": "meningioma",
    "meningioma_tumor": "meningioma",
    "meningioma tumor": "meningioma",
    "pituitary": "pituitary",
    "pituitary_tumor": "pituitary",
    "pituitary tumor": "pituitary",
    "notumor": "notumor",
    "no_tumor": "notumor",
    "no tumor": "notumor",
    "no-tumor": "notumor",
    "normal": "notumor",
}
CLASS_NAMES = ["glioma", "meningioma", "notumor", "pituitary"]
CLASS_TO_IDX = {name: i for i, name in enumerate(CLASS_NAMES)}
VALID_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

def normalized_name(name):
    return name.strip().lower().replace("-", "_")

def has_split_dirs(path):
    if not path.is_dir():
        return False
    children = {p.name.lower() for p in path.iterdir() if p.is_dir()}
    return bool({"training", "testing"} <= children or {"train", "test"} <= children)

def locate_dataset():
    if DATASET_PATH.strip():
        p = Path(DATASET_PATH.strip())
        if p.is_file() and p.suffix.lower() == ".zip":
            return p
        if p.is_dir():
            return p
        raise FileNotFoundError(f"DATASET_PATH does not exist: {p}")

    # Prefer already-extracted folders with Training/Testing or train/test.
    candidates = []
    for root in SEARCH_ROOTS:
        if root.exists():
            for current, dirs, files in os.walk(root):
                dirs[:] = [d for d in dirs if d not in {".git", ".cache"}]
                p = Path(current)
                if has_split_dirs(p):
                    candidates.append(p)
                # Avoid exploring very deep irrelevant directories.
                if len(p.parts) > len(root.parts) + 6:
                    dirs[:] = []

    if candidates:
        candidates.sort(key=lambda p: (len(p.parts), str(p)))
        return candidates[0]

    # Look for likely ZIP files.
    zips = []
    for root in SEARCH_ROOTS:
        if root.exists():
            zips.extend(root.glob("*.zip"))
            zips.extend(root.glob("**/*brain*tumor*.zip"))
            zips.extend(root.glob("**/*Brain*Tumor*.zip"))
    zips = list(dict.fromkeys(zips))
    if zips:
        return zips[0]

    raise FileNotFoundError(
        "Dataset folder/ZIP not found automatically. Set DATASET_PATH at the top "
        "to your extracted dataset folder or ZIP path."
    )

dataset_path = locate_dataset()
print(f"Dataset found: {dataset_path}")

# ------------------------------ EXTRACT ZIP ----------------------------
if dataset_path.is_file() and dataset_path.suffix.lower() == ".zip":
    extract_dir = Path("/content/NeuroScan_BrainMRI")
    extract_dir.mkdir(parents=True, exist_ok=True)
    print("Extracting ZIP; this may take a few minutes...")
    with zipfile.ZipFile(dataset_path, "r") as zf:
        zf.extractall(extract_dir)
    search_base = extract_dir
else:
    search_base = dataset_path

# Find the directory that directly contains the train/test split folders.
split_root_candidates = []
for current, dirs, files in os.walk(search_base):
    p = Path(current)
    children = {d.lower() for d in dirs}
    if {"training", "testing"} <= children or {"train", "test"} <= children:
        split_root_candidates.append(p)
    if len(p.parts) > len(search_base.parts) + 7:
        dirs[:] = []

if not split_root_candidates:
    raise FileNotFoundError(
        f"No Training/Testing or train/test folders found under {search_base}. "
        "Check the ZIP/folder structure and set DATASET_PATH correctly."
    )

split_root_candidates.sort(key=lambda p: (len(p.parts), str(p)))
SPLIT_ROOT = split_root_candidates[0]
print(f"Split root: {SPLIT_ROOT}")

root_children = {p.name.lower(): p for p in SPLIT_ROOT.iterdir() if p.is_dir()}
TRAIN_DIR = root_children.get("training", root_children.get("train"))
TEST_DIR = root_children.get("testing", root_children.get("test"))

if TRAIN_DIR is None or TEST_DIR is None:
    raise RuntimeError("Training or Testing folder is missing.")

# ----------------------- SCAN AND VALIDATE IMAGES ---------------------
def scan_images(split_dir, split_name):
    records, invalid = [], []
    for class_dir in sorted([p for p in split_dir.iterdir() if p.is_dir()]):
        raw_name = class_dir.name.strip().lower()
        key = normalized_name(raw_name)
        canonical = CLASS_ALIASES.get(key, CLASS_ALIASES.get(raw_name))
        if canonical is None:
            print(f"Warning: skipping unrecognized class folder: {class_dir.name}")
            continue

        for file_path in class_dir.rglob("*"):
            if not file_path.is_file() or file_path.suffix.lower() not in VALID_EXTS:
                continue
            try:
                with Image.open(file_path) as im:
                    im.verify()
                with Image.open(file_path) as im:
                    if im.width < 16 or im.height < 16:
                        raise ValueError("Image dimensions too small")
                    im.convert("RGB")
                records.append({
                    "path": str(file_path),
                    "class_name": canonical,
                    "label": CLASS_TO_IDX[canonical],
                    "split": split_name,
                    "width": im.width,
                    "height": im.height,
                })
            except Exception as e:
                invalid.append((str(file_path), str(e)))

    return records, invalid

train_records, invalid_train = scan_images(TRAIN_DIR, "train_source")
test_records, invalid_test = scan_images(TEST_DIR, "test")

print(f"Valid training-source images: {len(train_records):,}")
print(f"Valid testing images:         {len(test_records):,}")
print(f"Invalid/corrupt training images: {len(invalid_train):,}")
print(f"Invalid/corrupt testing images:  {len(invalid_test):,}")

if not train_records or not test_records:
    raise RuntimeError("No valid images found. Inspect the dataset folder names and structure.")

# ------------------------- DUPLICATE DETECTION -------------------------
# Exact file-content hashes. Keep the original official test split separate:
# remove exact duplicates from test if the same image exists in training.
def file_hash(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

print("Checking exact duplicates (this can take a little time)...")
train_hashes = {}
for rec in train_records:
    try:
        train_hashes.setdefault(file_hash(rec["path"]), rec["path"])
    except OSError:
        pass

clean_test_records = []
test_duplicate_paths = []
seen_test_hashes = set()

for rec in test_records:
    try:
        h = file_hash(rec["path"])
        if h in train_hashes or h in seen_test_hashes:
            test_duplicate_paths.append(rec["path"])
            continue
        seen_test_hashes.add(h)
        clean_test_records.append(rec)
    except OSError:
        continue

print(f"Exact duplicate test images removed: {len(test_duplicate_paths):,}")
test_records = clean_test_records

# Remove duplicate files within the training source before splitting.
unique_train_records, seen_train_hashes = [], set()
for rec in train_records:
    try:
        h = file_hash(rec["path"])
        if h not in seen_train_hashes:
            unique_train_records.append(rec)
            seen_train_hashes.add(h)
    except OSError:
        continue
train_records = unique_train_records

# ----------------------- STRATIFIED VALIDATION SPLIT -------------------
train_df_all = pd.DataFrame(train_records)
test_df = pd.DataFrame(test_records)

# Stratification needs at least 2 examples in every class.
counts = train_df_all["label"].value_counts()
if (counts < 2).any():
    raise ValueError("Some classes have fewer than 2 training images; cannot stratify the split.")

train_df, val_df = train_test_split(
    train_df_all,
    test_size=VAL_SIZE,
    random_state=SEED,
    stratify=train_df_all["label"]
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("\nFinal dataset sizes")
print(f"Train:      {len(train_df):,}")
print(f"Validation: {len(val_df):,}")
print(f"Test:       {len(test_df):,}")

# ------------------------------ TRANSFORMS -----------------------------
# Augmentation is applied ONLY to training images.
train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.ColorJitter(brightness=0.10, contrast=0.10, saturation=0.05),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

class BrainMRIDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.df = dataframe.reset_index(drop=True).copy()
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):
        row = self.df.iloc[index]
        with Image.open(row["path"]) as im:
            image = im.convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, int(row["label"])

train_dataset = BrainMRIDataset(train_df, train_transform)
val_dataset = BrainMRIDataset(val_df, eval_transform)
test_dataset = BrainMRIDataset(test_df, eval_transform)

# DataLoaders are ready for training. num_workers=2 is usually suitable in Colab.
train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available()
)

# ----------------------------- DATA SUMMARY ----------------------------
print("\nClass mapping (save this with the model weights):")
print(CLASS_TO_IDX)

print("\nTraining class distribution:")
print(train_df["class_name"].value_counts().reindex(CLASS_NAMES, fill_value=0))

print("\nValidation class distribution:")
print(val_df["class_name"].value_counts().reindex(CLASS_NAMES, fill_value=0))

print("\nTest class distribution:")
print(test_df["class_name"].value_counts().reindex(CLASS_NAMES, fill_value=0))

# --------------------------- VISUALIZE SAMPLES -------------------------
def denormalize(tensor):
    mean = torch.tensor(IMAGENET_MEAN).view(3, 1, 1)
    std = torch.tensor(IMAGENET_STD).view(3, 1, 1)
    return (tensor.cpu() * std + mean).clamp(0, 1)

fig, axes = plt.subplots(2, 4, figsize=(14, 7))
for ax in axes.flat:
    ax.axis("off")

sample_count = min(8, len(train_dataset))
for i in range(sample_count):
    image_tensor, label = train_dataset[i]  # train augmentation shown for illustration
    image_np = denormalize(image_tensor).permute(1, 2, 0).numpy()
    ax = axes.flat[i]
    ax.imshow(image_np)
    ax.set_title(CLASS_NAMES[label])
    ax.axis("off")

plt.suptitle("NeuroScan AI — Preprocessed Training Samples", fontsize=14)
plt.tight_layout()
plt.show()

# ------------------------- VERIFY A BATCH ------------------------------
images, labels = next(iter(train_loader))
print("\nPreprocessing verification:")
print(f"Image batch shape: {tuple(images.shape)}  # expected [batch, 3, 224, 224]")
print(f"Label batch shape: {tuple(labels.shape)}")
print(f"Pixel tensor dtype: {images.dtype}")
print(f"Device available: {'GPU' if torch.cuda.is_available() else 'CPU'}")

# ----------------------------- SAVE METADATA ---------------------------
OUTPUT_DIR = Path("/content/NeuroScan_preprocessing")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

train_df.to_csv(OUTPUT_DIR / "train_manifest.csv", index=False)
val_df.to_csv(OUTPUT_DIR / "val_manifest.csv", index=False)
test_df.to_csv(OUTPUT_DIR / "test_manifest.csv", index=False)

metadata = {
    "class_names": CLASS_NAMES,
    "class_to_idx": CLASS_TO_IDX,
    "image_size": IMAGE_SIZE,
    "imagenet_mean": list(IMAGENET_MEAN),
    "imagenet_std": list(IMAGENET_STD),
    "seed": SEED,
    "train_count": len(train_df),
    "validation_count": len(val_df),
    "test_count": len(test_df),
}
pd.Series(metadata).to_json(OUTPUT_DIR / "preprocessing_metadata.json", indent=2)

print(f"\nPreprocessing complete. Manifests and metadata saved in: {OUTPUT_DIR}")
print("Ready for model training: train_loader, val_loader, test_loader")
```

```python

# ============================================================
# NeuroScan AI — EfficientNet-B0 Training Pipeline
# Run AFTER your preprocessing cell in Google Colab
# ============================================================

import os
import json
import copy
import time
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, classification_report, confusion_matrix,
    ConfusionMatrixDisplay
)

from pathlib import Path

# -------------------- CONFIGURATION --------------------
SEED = 42
EPOCHS = 15
LEARNING_RATE = 1e-4
WEIGHT_DECAY = 1e-4
PATIENCE = 4
NUM_CLASSES = len(CLASS_NAMES)

OUTPUT_DIR = Path("/content/NeuroScan_AI_Results")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
if device.type == "cpu":
    print("WARNING: Enable a GPU in Colab: Runtime > Change runtime type > T4 GPU")

# -------------------- MODEL --------------------
# Use pretrained ImageNet weights for transfer learning.
try:
    weights = models.EfficientNet_B0_Weights.DEFAULT
    model = models.efficientnet_b0(weights=weights)
except Exception as e:
    print("Could not load pretrained weights:", e)
    print("Falling back to model initialization without pretrained weights.")
    model = models.efficientnet_b0(weights=None)

in_features = model.classifier[1].in_features
model.classifier[1] = nn.Linear(in_features, NUM_CLASSES)
model = model.to(device)

# Cross-entropy with inverse-frequency class weights.
train_counts = np.bincount(
    train_loader.dataset.df["label"].to_numpy(),
    minlength=NUM_CLASSES
)
class_weights = len(train_loader.dataset) / (
    NUM_CLASSES * np.maximum(train_counts, 1)
)
criterion = nn.CrossEntropyLoss(
    weight=torch.tensor(class_weights, dtype=torch.float32, device=device)
)

optimizer = optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="max", factor=0.5, patience=2
)

# AMP speeds up training on compatible GPUs.
use_amp = device.type == "cuda"
scaler = torch.amp.GradScaler("cuda", enabled=use_amp)

# -------------------- TRAIN / VALIDATE FUNCTIONS --------------------
def run_epoch(loader, training=True):
    model.train(training)
    total_loss = 0.0
    all_labels, all_preds = [], []

    for images, labels in loader:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        if training:
            optimizer.zero_grad(set_to_none=True)

        with torch.set_grad_enabled(training):
            with torch.autocast(
                device_type=device.type,
                dtype=torch.float16,
                enabled=use_amp
            ):
                outputs = model(images)
                loss = criterion(outputs, labels)

            if training:
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()

        total_loss += loss.item() * images.size(0)
        preds = outputs.argmax(dim=1)

        all_labels.extend(labels.detach().cpu().numpy())
        all_preds.extend(preds.detach().cpu().numpy())

    avg_loss = total_loss / len(loader.dataset)
    acc = accuracy_score(all_labels, all_preds)

    return avg_loss, acc, np.array(all_labels), np.array(all_preds)

# -------------------- TRAINING LOOP --------------------
history = []
best_val_acc = -1.0
best_val_loss = float("inf")
best_epoch = 0
best_state = None
epochs_without_improvement = 0

start_time = time.time()

for epoch in range(1, EPOCHS + 1):
    epoch_start = time.time()

    train_loss, train_acc, _, _ = run_epoch(train_loader, training=True)

    with torch.no_grad():
        val_loss, val_acc, _, _ = run_epoch(val_loader, training=False)

    scheduler.step(val_acc)

    current_lr = optimizer.param_groups[0]["lr"]
    epoch_time = time.time() - epoch_start

    record = {
        "epoch": epoch,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "val_loss": val_loss,
        "val_accuracy": val_acc,
        "learning_rate": current_lr,
        "epoch_seconds": epoch_time
    }
    history.append(record)

    print(
        f"Epoch [{epoch:02d}/{EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4%} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4%} | "
        f"LR: {current_lr:.2e} | "
        f"Time: {epoch_time:.1f}s"
    )

    # Save best checkpoint based on validation accuracy.
    improved = (
        val_acc > best_val_acc
        or (val_acc == best_val_acc and val_loss < best_val_loss)
    )

    if improved:
        best_val_acc = val_acc
        best_val_loss = val_loss
        best_epoch = epoch
        best_state = copy.deepcopy(model.state_dict())
        epochs_without_improvement = 0

        torch.save({
            "model_state_dict": best_state,
            "class_names": CLASS_NAMES,
            "class_to_idx": CLASS_TO_IDX,
            "num_classes": NUM_CLASSES,
            "image_size": IMAGE_SIZE,
            "mean": list(IMAGENET_MEAN),
            "std": list(IMAGENET_STD),
            "architecture": "efficientnet_b0",
            "best_epoch": best_epoch,
            "best_val_accuracy": best_val_acc,
            "best_val_loss": best_val_loss,
        }, OUTPUT_DIR / "best_model.pth")

        print("  -> Best model checkpoint saved.")
    else:
        epochs_without_improvement += 1

    if epochs_without_improvement >= PATIENCE:
        print(f"Early stopping at epoch {epoch}.")
        break

training_seconds = time.time() - start_time

# -------------------- SAVE TRAINING HISTORY --------------------
history_df = pd.DataFrame(history)
history_df.to_csv(OUTPUT_DIR / "training_history.csv", index=False)

with open(OUTPUT_DIR / "training_config.json", "w") as f:
    json.dump({
        "architecture": "efficientnet_b0",
        "epochs_requested": EPOCHS,
        "epochs_completed": len(history),
        "best_epoch": best_epoch,
        "best_validation_accuracy": best_val_acc,
        "best_validation_loss": best_val_loss,
        "training_time_seconds": training_seconds,
        "class_names": CLASS_NAMES,
        "class_to_idx": CLASS_TO_IDX,
        "image_size": IMAGE_SIZE,
        "mean": list(IMAGENET_MEAN),
        "std": list(IMAGENET_STD),
        "seed": SEED
    }, f, indent=2)

# -------------------- RESTORE BEST MODEL --------------------
checkpoint = torch.load(
    OUTPUT_DIR / "best_model.pth",
    map_location=device,
    weights_only=True
)
model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

print(f"\nBest epoch: {best_epoch}")
print(f"Best validation accuracy: {best_val_acc:.4%}")
print(f"Training time: {training_seconds / 60:.2f} minutes")

# -------------------- FINAL TEST EVALUATION --------------------
all_test_labels, all_test_preds, all_test_probs = [], [], []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device, non_blocking=True)

        with torch.autocast(
            device_type=device.type,
            dtype=torch.float16,
            enabled=use_amp
        ):
            logits = model(images)

        probs = torch.softmax(logits.float(), dim=1)
        preds = probs.argmax(dim=1)

        all_test_labels.extend(labels.numpy())
        all_test_preds.extend(preds.cpu().numpy())
        all_test_probs.extend(probs.cpu().numpy())

y_true = np.array(all_test_labels)
y_pred = np.array(all_test_preds)

test_accuracy = accuracy_score(y_true, y_pred)
test_precision = precision_score(
    y_true, y_pred, average="macro", zero_division=0
)
test_recall = recall_score(
    y_true, y_pred, average="macro", zero_division=0
)
test_f1 = f1_score(
    y_true, y_pred, average="macro", zero_division=0
)

print("\n========== FINAL TEST RESULTS ==========")
print(f"Test Accuracy:  {test_accuracy:.4%}")
print(f"Macro Precision:{test_precision:.4%}")
print(f"Macro Recall:   {test_recall:.4%}")
print(f"Macro F1-score: {test_f1:.4%}")

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    target_names=CLASS_NAMES,
    zero_division=0
))

metrics = {
    "test_accuracy": float(test_accuracy),
    "macro_precision": float(test_precision),
    "macro_recall": float(test_recall),
    "macro_f1": float(test_f1),
    "best_epoch": int(best_epoch),
    "best_validation_accuracy": float(best_val_acc),
    "training_time_seconds": float(training_seconds)
}

with open(OUTPUT_DIR / "test_metrics.json", "w") as f:
    json.dump(metrics, f, indent=2)

# -------------------- CONFUSION MATRIX --------------------
cm = confusion_matrix(
    y_true, y_pred, labels=list(range(NUM_CLASSES))
)
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm, display_labels=CLASS_NAMES
)
disp.plot(cmap="Blues", xticks_rotation=30, values_format="d")
plt.title("NeuroScan AI — Test Confusion Matrix")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "confusion_matrix.png", dpi=200)
plt.show()

# -------------------- TRAINING CURVES --------------------
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].plot(history_df["epoch"], history_df["train_loss"], label="Train Loss")
axes[0].plot(history_df["epoch"], history_df["val_loss"], label="Validation Loss")
axes[0].set_title("Loss")
axes[0].set_xlabel("Epoch")
axes[0].legend()
axes[0].grid(True)

axes[1].plot(
    history_df["epoch"], history_df["train_accuracy"], label="Train Accuracy"
)
axes[1].plot(
    history_df["epoch"], history_df["val_accuracy"], label="Validation Accuracy"
)
axes[1].set_title("Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.savefig(OUTPUT_DIR / "training_curves.png", dpi=200)
plt.show()

# -------------------- EXPORT STANDALONE WEIGHTS --------------------
# This is convenient for inference scripts that expect a state_dict.
torch.save(
    model.state_dict(),
    OUTPUT_DIR / "best_model_state_dict.pth"
)

print("\n========== SAVED FILES ==========")
for p in sorted(OUTPUT_DIR.iterdir()):
    if p.is_file():
        print(f"{p.name}: {p.stat().st_size / (1024 * 1024):.2f} MB")

print("\nAll training artifacts saved in:", OUTPUT_DIR)
print("Next: download the output folder or create a ZIP for GitHub.")
```

```python

import os
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np

from PIL import Image
from torchvision import models, transforms
from google.colab import files

# 1. Load the best saved model
MODEL_PATH = "/content/NeuroScan_AI_Results/best_model.pth"

if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(
        "Model not found. Check that best_model.pth exists in "
        "/content/NeuroScan_AI_Results/"
    )

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

checkpoint = torch.load(
    MODEL_PATH,
    map_location=device,
    weights_only=True
)

class_names = checkpoint["class_names"]
image_size = checkpoint.get("image_size", 224)
mean = checkpoint.get("mean", [0.485, 0.456, 0.406])
std = checkpoint.get("std", [0.229, 0.224, 0.225])

# 2. Rebuild exactly the same model architecture
model = models.efficientnet_b0(weights=None)
in_features = model.classifier[1].in_features
model.classifier[1] = nn.Linear(in_features, len(class_names))

model.load_state_dict(checkpoint["model_state_dict"])
model = model.to(device)
model.eval()

# 3. Use the same preprocessing as evaluation
preprocess = transforms.Compose([
    transforms.Resize((image_size, image_size)),
    transforms.ToTensor(),
    transforms.Normalize(mean=mean, std=std)
])

# 4. Upload an MRI image from your computer
print("Upload a brain MRI image (JPG, JPEG, or PNG):")
uploaded = files.upload()

if not uploaded:
    raise ValueError("No image uploaded.")

filename = next(iter(uploaded))

try:
    image = Image.open(filename).convert("RGB")
except Exception as e:
    raise ValueError(f"Could not open uploaded image: {e}")

input_tensor = preprocess(image).unsqueeze(0).to(device)

# 5. Run prediction
with torch.no_grad():
    logits = model(input_tensor)
    probabilities = torch.softmax(logits, dim=1)[0]

predicted_index = probabilities.argmax().item()
predicted_class = class_names[predicted_index]
confidence = probabilities[predicted_index].item() * 100

# 6. Display the image and prediction results
plt.figure(figsize=(6, 6))
plt.imshow(image)
plt.title(
    f"Prediction: {predicted_class}\n"
    f"Model confidence: {confidence:.2f}%"
)
plt.axis("off")
plt.show()

print("=" * 40)
print("NEUROSCAN AI — PREDICTION RESULTS")
print("=" * 40)
print(f"Predicted class: {predicted_class}")
print(f"Confidence: {confidence:.2f}%")

print("\nProbability for each class:")
for i in torch.argsort(probabilities, descending=True).tolist():
    print(f"{class_names[i]}: {probabilities[i].item() * 100:.2f}%")

print("\nNote: This is an academic image-classification model,")
print("not a medical diagnosis.")
```
